```python
def extract_id(file_path):
    file_name = os.path.basename(file_path)
    base_name = os.path.splitext(file_name)[0]  # remove .mp3

    if base_name.startswith('XC'):
        return base_name.split(' ')[0]  # take 'XC116651' before the first space
    else:
        return base_name  # just the numeric ID

df["File ID"] = df["File"].apply(extract_id)
# # add and drop columns
df = df.drop(columns=['File'])

df["Call"] = None
df["Buzz"] = None
```
