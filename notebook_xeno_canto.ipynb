```python
import pandas as pd
```

```python
df = pd.read_csv('birdnet_xeno_canto/xeno_canto.csv')
```

```python

# extract XC file name:XC414415
df["File Name"] = df["File"].apply(lambda x: 'XC' + x.split('_XC')[1].split('.')[0] if '_XC' in x else None)

# add and drop columns
df = df.drop(columns=['File'])
df["Latitude"] = None
df["Longitude"] = None
df["Elevation"] = None
df["Year"] = None
df["Month"] = None
df["Day"] = None
df["Time"] = None
df["Buzz"] = None

# save the preocessed dataset
df.to_csv("birdnet_xeno_canto/processed_xeno_canto.csv", index=False)

print(df.head())
```
