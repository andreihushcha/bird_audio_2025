```python
import os
import pandas as pd
```

```python

def prefix_filenames(root_dir):
    """
    For each file inside a subfolder, renames it by adding the folder name as a prefix.
    Example: 'bird001.wav' in folder 'psh2' becomes 'psh2_bird001.wav'
    """
    for folder_name in os.listdir(root_dir):
        folder_path = os.path.join(root_dir, folder_name)
        
        if os.path.isdir(folder_path):
            for filename in os.listdir(folder_path):
                file_path = os.path.join(folder_path, filename)
                
                if os.path.isfile(file_path):
                    # Skip files already renamed
                    if filename.startswith(folder_name + "_"):
                        continue

                    new_filename = f"{folder_name}_{filename}"
                    new_filepath = os.path.join(folder_path, new_filename)
                    
                    os.rename(file_path, new_filepath)
                    print(f"Renamed: {filename} → {new_filename}")


```
